## Linear Regression Algorithm and Logistic Regression Algorithm

### Aim

To practically implement and understand the fundamental concepts of Linear Regression and Logistic Regression, two widely used supervised machine learning algorithms, through hands-on coding and data analysis.

### Objectives

- To understand the theory and mathematical formulation of Linear Regression for continuous target variables.
- To understand the application and principles of Logistic Regression for categorical target variables.
- To gain practical experience in implementing both algorithms using Python and relevant libraries.
- To learn how to prepare data, train models, and evaluate their performance.

### Relevance

Linear Regression is used to predict continuous numeric values such as price, temperature, or sales using Ordinary Least Squares.

Logistic Regression is used for classification problems, predicting the probability of categories such as yes/no or 0/1 using the sigmoid function.

### Step 1: Import Libraries

The required libraries are imported for numerical operations, data handling, data splitting, model training, evaluation, and visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, accuracy_score
```

### Step 2: Generate Sample Data

For Linear Regression, sample data is generated with a linear relationship between the input variable `X` and continuous target variable `y`.

```python
np.random.seed(0)

X_linear = 2 * np.random.rand(100, 1)
y_linear = 4 + 3 * X_linear + np.random.randn(100, 1)
```

For Logistic Regression, random input data is generated and converted into binary classes. Values greater than 0 belong to class 1, while the remaining values belong to class 0.

```python
X_logistic = np.random.randn(100, 1)
y_logistic = (X_logistic > 0).astype(int)
```

### Step 3: Data Splitting

The Linear Regression dataset is divided into training and testing sets. 80% of the data is used for training and 20% for testing.

```python
X_train_linear, X_test_linear, y_train_linear, y_test_linear = train_test_split(
    X_linear,
    y_linear,
    test_size=0.2,
    random_state=42
)
```

The Logistic Regression dataset is also divided into 80% training data and 20% testing data.

```python
X_train_logistic, X_test_logistic, y_train_logistic, y_test_logistic = train_test_split(
    X_logistic,
    y_logistic,
    test_size=0.2,
    random_state=42
)
```

### Step 4: Model Training and Prediction

The Linear Regression model is trained using the training data and then used to predict continuous values for the test data.

```python
linear_model = LinearRegression()

linear_model.fit(X_train_linear, y_train_linear)

y_pred_linear = linear_model.predict(X_test_linear)
```

The Logistic Regression model is trained using the training data and used to predict the binary class labels of the test data.

```python
logistic_model = LogisticRegression()

logistic_model.fit(X_train_logistic, y_train_logistic.ravel())

y_pred_logistic = logistic_model.predict(X_test_logistic)
```

### Step 5: Model Evaluation

Mean Squared Error (MSE) measures the average squared difference between the actual and predicted values. A lower MSE indicates smaller prediction errors.

```python
mse = mean_squared_error(y_test_linear, y_pred_linear)

print(f"Linear Regression Mean Squared Error: {mse:.2f}")
```

Accuracy measures the proportion of correctly classified samples out of all test samples.

```python
accuracy = accuracy_score(y_test_logistic, y_pred_logistic)

print(f"Logistic Regression Accuracy: {accuracy:.2f}")
```

### Step 6: Visualization

The results of Linear Regression and Logistic Regression are visualized to compare the actual and predicted values.

```python
plt.figure(figsize=(10, 5))

# Linear Regression
plt.subplot(1, 2, 1)

plt.scatter(
    X_test_linear,
    y_test_linear,
    color='blue',
    label='Actual'
)

plt.plot(
    X_test_linear,
    y_pred_linear,
    color='red',
    label='Predicted'
)

plt.title('Linear Regression')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()


# Logistic Regression
plt.subplot(1, 2, 2)

plt.scatter(
    X_test_logistic,
    y_test_logistic,
    color='green',
    label='Actual'
)

plt.scatter(
    X_test_logistic,
    y_pred_logistic,
    color='purple',
    marker='x',
    label='Predicted'
)

plt.title('Logistic Regression')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

plt.tight_layout()
plt.show()
```

## Conclusion

We successfully implemented and evaluated Linear Regression and Logistic Regression. Linear Regression was used to predict a continuous target variable by fitting a straight line that minimizes prediction error. Logistic Regression was used for binary classification by predicting class labels. The models were trained, tested, evaluated, and visualized using Python and Scikit-learn.
