### Step 6: Outlier Detection

A separate copy of the original Employee dataset is loaded for comparison during the outlier analysis.

```python
df1 = pd.read_csv('Employee.csv')

df1.describe()
```

The first quartile (Q1) represents the 25th percentile of the Age column.

```python
q1 = df['Age'].quantile(0.25)

q1
```

The third quartile (Q3) represents the 75th percentile of the Age column.

```python
q3 = df['Age'].quantile(0.75)

q3
```

The Interquartile Range (IQR) is calculated as the difference between Q3 and Q1. It represents the spread of the middle 50% of the data.

```python
iqr = q3 - q1

iqr
```

The lower limit is calculated using the IQR rule. Values below this limit can be considered potential outliers.

```python
lower = q1 - 1.5 * iqr

lower
```

The upper limit is calculated using the IQR rule. Values above this limit can be considered potential outliers.

```python
upper = q3 + 1.5 * iqr

upper
```

This checks for Age values outside the calculated limits.

```python
df[(df['Age'] < lower) & (df['Age'] > upper)]
```

The first quartile (Q1) of the Salary column is calculated to begin the IQR-based outlier detection.

```python
q1 = df['Salary'].quantile(0.25)

q1
```

The third quartile (Q3) of the Salary column is calculated.

```python
q3 = df['Salary'].quantile(0.75)

q3
```

The IQR for Salary is calculated as Q3 minus Q1.

```python
iqr = q3 - q1

iqr
```

The lower limit for Salary is calculated using the IQR rule.

```python
lower = q1 - 1.5 * iqr

lower
```

The upper limit for Salary is calculated using the IQR rule.

```python
upper = q3 + 1.5 * iqr

upper
```

This checks for Salary values outside the calculated limits.

```python
df[(df['Salary'] < lower) & (df['Salary'] > upper)]
```

### Step 7: Handling Missing Values

Before replacing missing values, we check the number of missing values in each column.

```python
df.isnull().sum()
```

The missing values in the Age column are replaced with the mean age. This allows us to retain the rows instead of deleting them.

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)

df['Age'].isnull().sum()
```

The missing values in the Salary column are replaced with the mean salary.

```python
df['Salary'].fillna(df['Salary'].mean(), inplace=True)

df['Salary'].isnull().sum()
```

The missing values in the Place column are replaced with the mode, which is the most frequently occurring place.

```python
df['Place'].fillna(df['Place'].mode()[0], inplace=True)

df['Place'].isnull().sum()
```

The missing values in the Gender column are replaced using the mode because Gender is a categorical variable.

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)

df['Gender'].isnull().sum()
```

Finally, we check the complete dataset again to confirm whether any missing values remain.

```python
df.isnull().sum()
```

### Step 8: Encoding Categorical Variables

Before encoding, we check the data types of all columns. Machine learning algorithms generally require numerical input, so categorical columns need to be converted into numerical form.

```python
df.dtypes
```

The Gender column contains categorical values. We convert the categories into numerical values using one-hot encoding.

```python
df = pd.get_dummies(df, columns=['Gender'], dtype=int)

df.head()
```

The Place column is also categorical. One-hot encoding converts each unique place into a separate numerical column.

```python
df = pd.get_dummies(df, columns=['Place'], dtype=int)

df.head()
```

The Company column contains categorical company names. One-hot encoding converts these categories into numerical columns.

```python
df = pd.get_dummies(df, columns=['Company'], dtype=int)

df.head()
```

The Country column is categorical, so it is converted into numerical columns using one-hot encoding.

```python
df = pd.get_dummies(df, columns=['Country'], dtype=int)

df.head()
```

The data types are checked again to verify that the categorical columns have been converted into numerical form.

```python
df.dtypes
```

### Step 9: Feature Scaling

Standardization changes the numerical features so that they have a mean of approximately 0 and a standard deviation of approximately 1.

```python
from sklearn.preprocessing import StandardScaler
```

A StandardScaler object is created to perform standardization on the numerical data.

```python
scaler = StandardScaler()
```

The scaler is fitted to the dataset and the numerical values are transformed into standardized values.

```python
df_standardized = scaler.fit_transform(df)

df_standardized
```

The standardized NumPy array is converted back into a Pandas DataFrame so that the data can be easily viewed and analyzed.

```python
df_standardized = pd.DataFrame(
    df_standardized,
    columns=df.columns
)

df_standardized.head()
```

### Step 10: Min-Max Normalization

Min-Max normalization transforms the values into a fixed range, usually between 0 and 1.

```python
from sklearn.preprocessing import MinMaxScaler
```

A MinMaxScaler object is created to normalize the dataset.

```python
minmax_scaler = MinMaxScaler()
```

The MinMaxScaler is fitted to the dataset and transforms the values into the range from 0 to 1.

```python
df_normalized = minmax_scaler.fit_transform(df)

df_normalized
```

The normalized data is converted into a Pandas DataFrame for easier viewing.

```python
df_normalized = pd.DataFrame(
    df_normalized,
    columns=df.columns
)

df_normalized.head()
```
